```python
import os
import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt
```

```python
DATASET_PATH = "brats2021/BraTS2021_Training_Data"

patients = sorted(os.listdir(DATASET_PATH))
print("Number of patients:", len(patients))

patient_id = patients[0]
patient_path = os.path.join(DATASET_PATH, patient_id)

print("Selected patient:", patient_id)
print(os.listdir(patient_path))
```

```python
def load_nifti(path):
    nii = nib.load(path)
    return nii.get_fdata(), nii.affine, nii.header

flair, _, _ = load_nifti(os.path.join(patient_path, f"{patient_id}_flair.nii.gz"))
t1, _, _    = load_nifti(os.path.join(patient_path, f"{patient_id}_t1.nii.gz"))
t1ce, _, _  = load_nifti(os.path.join(patient_path, f"{patient_id}_t1ce.nii.gz"))
t2, _, _    = load_nifti(os.path.join(patient_path, f"{patient_id}_t2.nii.gz"))
seg, affine, header = load_nifti(os.path.join(patient_path, f"{patient_id}_seg.nii.gz"))
```
